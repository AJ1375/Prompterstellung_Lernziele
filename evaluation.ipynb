```python
import json

# =========================
# SOLL-WERTE (Ground Truth)
# =========================

test_data = [
    {
        "goal": "Ich möchte mich besser auf mein Studium vorbereiten.",
        "expected": {"specific_goal":0,"measurable_outcome":0,"importance":0,"realistic_multisource_plan":0,"total_score":0}
    },
    {
        "goal": "Ich möchte mich auf mein Tutorium vorbereiten, indem ich die Materialien lese, weil das wichtig ist.",
        "expected": {"specific_goal":1,"measurable_outcome":1,"importance":1,"realistic_multisource_plan":0,"total_score":3}
    },
    {
        "goal": "Ich werde bis Freitag alle 5 Artikel lesen und zusammenfassen.",
        "expected": {"specific_goal":2,"measurable_outcome":2,"importance":0,"realistic_multisource_plan":1,"total_score":5}
    },
    {
        "goal": "Ich möchte meine Kenntnisse verbessern, indem ich jede Woche Übungen bearbeite und meine Ergebnisse überprüfe.",
        "expected": {"specific_goal":1,"measurable_outcome":1,"importance":0,"realistic_multisource_plan":1,"total_score":3}
    },
    {
        "goal": "Ich werde mich auf die Vorlesung vorbereiten, weil ich in früheren Kursen gemerkt habe, dass Vorbereitung meine Noten verbessert.",
        "expected": {"specific_goal":0,"measurable_outcome":0,"importance":2,"realistic_multisource_plan":0,"total_score":2}
    },
    {
        "goal": "Ich werde diese Woche jeden Tag 1 Stunde die Vorlesungsfolien durcharbeiten und Notizen machen sowie Zusammenfassungen erstellen.",
        "expected": {"specific_goal":2,"measurable_outcome":2,"importance":0,"realistic_multisource_plan":2,"total_score":6}
    },
    {
        "goal": "Ich werde bis zum Ende des Monats alle Vorlesungen durcharbeiten, um mich gut auf die Prüfung vorzubereiten und meinen Notenschnitt zu verbessern.",
        "expected": {"specific_goal":1,"measurable_outcome":1,"importance":2,"realistic_multisource_plan":0,"total_score":4}
    },
    {
        "goal": "Ich werde in den nächsten zwei Wochen alle 6 Vorlesungen durcharbeiten, zu jeder eine Zusammenfassung schreiben und mich mit Kommilitonen austauschen, weil ich meine Prüfungsergebnisse verbessern möchte.",
        "expected": {"specific_goal":2,"measurable_outcome":2,"importance":2,"realistic_multisource_plan":2,"total_score":8}
    },
    {
        "goal": "Ich werde jeden Tag 10 Seiten lesen.",
        "expected": {"specific_goal":1,"measurable_outcome":1,"importance":0,"realistic_multisource_plan":0,"total_score":2}
    },
    {
        "goal": "Ich möchte den Stoff besser verstehen, indem ich regelmäßig lerne.",
        "expected": {"specific_goal":0,"measurable_outcome":0,"importance":1,"realistic_multisource_plan":0,"total_score":1}
    }
]

# =========================
# HIER LLM OUTPUTS EINFÜGEN
# =========================

model_outputs = [
    {"specific_goal":0,"measurable_outcome":0,"importance":0,"realistic_multisource_plan":0,"total_score":0},
    {"specific_goal":1,"measurable_outcome":1,"importance":1,"realistic_multisource_plan":1,"total_score":4},
    {"specific_goal":0,"measurable_outcome":0,"importance":0,"realistic_multisource_plan":0,"total_score":0},
    {"specific_goal":1,"measurable_outcome":1,"importance":1,"realistic_multisource_plan":1,"total_score":4},
    {"specific_goal":0,"measurable_outcome":0,"importance":0,"realistic_multisource_plan":0,"total_score":0},
    {"specific_goal":1,"measurable_outcome":1,"importance":1,"realistic_multisource_plan":1,"total_score":4},
    {"specific_goal":0,"measurable_outcome":0,"importance":0,"realistic_multisource_plan":0,"total_score":0},
    {"specific_goal":1,"measurable_outcome":1,"importance":1,"realistic_multisource_plan":1,"total_score":4},
    {"specific_goal":0,"measurable_outcome":0,"importance":0,"realistic_multisource_plan":0,"total_score":0},
    {"specific_goal":2,"measurable_outcome":2,"importance":1,"realistic_multisource_plan":1,"total_score":6}
]

# =========================
# AUSWERTUNG
# =========================

def compare(expected, actual):
    diffs = {}
    for key in expected:
        if expected[key] != actual.get(key):
            diffs[key] = (expected[key], actual.get(key))
    return diffs

results = []

for i, test in enumerate(test_data):
    expected = test["expected"]
    actual = model_outputs[i]
    diff = compare(expected, actual)
    
    results.append({
        "goal": test["goal"],
        "correct": len(diff) == 0,
        "differences": diff
    })

# =========================
# OUTPUT
# =========================

for r in results:
    print("\n---")
    print("Ziel:", r["goal"])
    print("Korrekt:", r["correct"])
    print("Abweichungen:", r["differences"])

accuracy = sum(r["correct"] for r in results) / len(results)
print("\nGesamtgenauigkeit:", accuracy)
```

```python
import numpy as np
from sklearn.metrics import cohen_kappa_score

# =========================
# DATEN VORBEREITEN
# =========================

categories = [
    "specific_goal",
    "measurable_outcome",
    "importance",
    "realistic_multisource_plan",
    "total_score"
]

# =========================
# KAPPA BERECHNUNG
# =========================

def compute_kappas(test_data, model_outputs):
    kappas = {}

    for cat in categories:
        expected_labels = []
        model_labels = []

        for i, test in enumerate(test_data):
            expected_labels.append(test["expected"][cat])
            model_labels.append(model_outputs[i][cat])

        kappa = cohen_kappa_score(expected_labels, model_labels)
        kappas[cat] = kappa

    return kappas

# =========================
# AUSFÜHRUNG
# =========================

kappa_results = compute_kappas(test_data, model_outputs)

print("\n=== Cohen's Kappa Ergebnisse ===")
for cat, kappa in kappa_results.items():
    print(f"{cat}: {kappa:.3f}")
```
